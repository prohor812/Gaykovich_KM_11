# Задание 9.1. Запись данных в текстовый файл

Сначала рассмотрим простой пример записи строк в текстовый файл.


```python
# имя файла
file_name = 'test.txt'

# открываем файл для записи
f = open(file_name,'w')

# создаём список строк
lines_list = [f'{x}\n' for x in 'test']

# записываем строки
f.writelines(lines_list)

# закрываем файл
f.close()
```

### Проверка содержимого файла

```python
# открываем файл для чтения
with open(file_name,'r') as f:

    # выводим строки файла
    for line in f:
        print(line,end='')
```

Для больших объёмов данных удобнее использовать генераторные функции.


```python
from time import process_time

# генератор строк
def lines_gen_fun(number_lines=1000):

    # размер записанных данных
    file_size = 0

    for x in range(number_lines):

        # создаём строку
        line = f'{x}\n'

        # возвращаем строку
        yield line

        # увеличиваем размер
        file_size += len(line)

    print(file_size/1024**2,'Mb')
```

### Измерение времени выполнения

```python
start = process_time()

list(lines_gen_fun())

end = process_time()

print(end-start,'секунд')
```

# Задание 9.2. Генерация строк

Согласно методичке необходимо создать строки со случайным количеством слов и случайным набором символов.


```python
from random import randint

# диапазон латинских символов
latin_code = (ord('a'),ord('z'))

# диапазон количества слов
number_words = (10,20)

# диапазон длины слова
number_symbols = (5,10)

# случайная длина слова
len_word = randint(*number_symbols)

# создаём слово
word = ''.join([chr(randint(*latin_code)) for i in range(len_word)])

print(word)
```

### Генераторная функция lines_gen_fun_v2

Функция создаёт строки до достижения заданного размера файла.


```python
from random import randint

def lines_gen_fun_v2(file_size,
                     type_symbols='latin',
                     number_words=(10,20),
                     number_symbols=(5,10)):
    '''Генерация строк заданного размера'''

    # выбираем набор символов
    if type_symbols == 'latin':
        symbols = (ord('a'),ord('z'))

    elif type_symbols == 'digits':
        symbols = (ord('0'),ord('9'))

    else:
        symbols = (ord('а'),ord('я'))

    # текущий размер файла
    current_size = 0

    while current_size < file_size*1024*1024:

        # количество слов в строке
        count_words = randint(*number_words)

        words = []

        for i in range(count_words):

            # длина слова
            len_word = randint(*number_symbols)

            # создаём слово
            word = ''.join(
                chr(randint(*symbols))
                for j in range(len_word)
            )

            words.append(word)

        # формируем строку
        line = ' '.join(words)+'\n'

        # увеличиваем размер
        current_size += len(line.encode('utf8'))

        # возвращаем строку
        yield line
```

### Проверка работы функции

```python
gen = lines_gen_fun_v2(1)

for i,line in enumerate(gen):

    print(line)

    if i == 2:
        break
```

# Задание 9.3. Создание функции file_gen

Функция должна создавать текстовый файл заданного размера и выводить информацию о процессе записи.


```python
from time import process_time

def file_gen(file_name,
             file_size,
             type_symbols='latin',
             number_words=(10,20),
             number_symbols=(5,10)):
    '''
    Создание текстового файла
    заданного размера
    '''

    # начало измерения времени
    start = process_time()

    # открываем файл
    with open(file_name,'w',encoding='utf8') as f:

        # генерация строк
        for line in lines_gen_fun_v2(file_size,
                                     type_symbols,
                                     number_words,
                                     number_symbols):

            # запись строки
            f.write(line)

    # окончание измерения времени
    end = process_time()

    # вывод времени
    print('Время выполнения:',end-start,'секунд')

    # вывод размера
    print('Размер файла:',file_size,'Mb')
```

### Тестирование функции file_gen

```python
file_gen('latin.txt',1,'latin')

file_gen('digits.txt',1,'digits')
```
