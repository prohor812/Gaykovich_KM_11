# Задание 7.1. Определение классов

Бинарное дерево поиска описывается с помощью трех классов: `BinaryTree`, `BinaryNode`, `EmptyNode`, связанных друг с другом на основе композиции.

## Класс EmptyNode

Класс `EmptyNode` определяет шаблон для создания объекта, представляющего «пустой» элемент (вершину) бинарного дерева поиска без значения и без поддеревьев. Экземпляры класса `EmptyNode` используются для представления пустого дерева и пустых поддеревьев.

Класс `EmptyNode` определяется двумя методами: `__repr__` и `insert`. Метод инициализации отсутствует, так как экземпляры класса `EmptyNode` не хранят никаких данных.

```python
class EmptyNode:                          # класс для представления пустой вершины бинарного дерева поиска
    def __repr__(self):                            # метод строкового представления экземпляра класса
        return '*'                                 # пустая вершина выводится символом *
    def insert(self, value):                       # метод вставки в пустую вершину нового элемента со значением value
        return BinaryNode(self, value, self)       # возвращаем новый экземпляр BinaryNode; левое и правое поддеревья — ссылки на пустую вершину (self)
```

## Класс BinaryTree

Класс `BinaryTree` определяет шаблон для создания объекта, представляющего бинарное дерево поиска.

Определение класса `BinaryTree` содержит три метода: `__init__`, `__repr__` и `insert`.

Экземпляр класса `BinaryTree` содержит атрибут `root`, который представляет корень дерева. Атрибут `root` хранит ссылку на экземпляр класса `EmptyNode`, если дерево пустое, или экземпляр класса `BinaryNode`, если дерево не пусто. При таком связывании классов реализуется концепция ООП — композиция.

```python
class BinaryTree:                         # класс для представления бинарного дерева поиска
    def __init__(self):                            # метод инициализации экземпляра класса
        self.root = EmptyNode()                    # атрибут root — ссылка на корень дерева; изначально дерево пусто, корень — экземпляр EmptyNode
    def __repr__(self):                            # метод строкового представления экземпляра класса
        return repr(self.root)                     # возвращаем строковое представление корня дерева
    def insert(self, value):                       # метод вставки нового элемента в дерево; новый элемент всегда является листом
        self.root = self.root.insert(value)        # метод insert реализуется через метод insert корневой вершины; атрибут root переопределяется ссылкой на новый объект
```

# Задание 7.2. Метод вставки элемента в бинарное дерево поиска

Переопределим класс `BinaryNode`, добавив метод `insert(self, value)`.

Алгоритм метода `insert(self, value)` класса `BinaryNode` определён рекурсивно:
- добавляемое значение `value` сравнивается со значением корня `self.value`;
- если `value < self.value`, то вызывается метод `insert(value)` для левого поддерева;
- если `value >= self.value`, то вызывается метод `insert(value)` для правого поддерева.

```python
class BinaryNode:                         # класс для представления вершины бинарного дерева поиска со значением и поддеревьями
    def __init__(self, left, value, right):        # метод инициализации экземпляра класса
        self.left = left                           # атрибут left — ссылка на корень левого поддерева
        self.value = value                         # атрибут value — значение вершины
        self.right = right                         # атрибут right — ссылка на корень правого поддерева
    def __repr__(self):                            # метод строкового представления экземпляра класса
        return f'({self.left}, {self.value}, {self.right})'  # возвращаем кортеж (левое поддерево, значение, правое поддерево)
    def insert(self, value):                       # метод вставки новой вершины со значением value
        if value < self.value:                     # если добавляемое значение меньше значения текущей вершины
            self.left = self.left.insert(value)    # рекурсивно вставляем в левое поддерево
        else:                                      # если добавляемое значение больше или равно значению текущей вершины
            self.right = self.right.insert(value)  # рекурсивно вставляем в правое поддерево
        return self                                # возвращаем ссылку на текущую вершину
```

### Пример 1: построение дерева из списка `list`

```python
source_data = [5, 1, 10, 3, 4]             # список числовых значений для построения дерева
tree = BinaryTree()                                # создаём пустое бинарное дерево поиска
for i in source_data:                              # последовательно добавляем элементы списка в дерево
    tree.insert(i)
print(tree)                                        # выводим дерево
```

### Пример 2: построение дерева из `str`

```python
tree_str = BinaryTree()                     # создаём пустое бинарное дерево поиска
for ch in 'binary':                                # добавляем символы строки в дерево
    tree_str.insert(ch)                            # для символов (str) определены операции сравнения
print(tree_str)
```

### Пример 3: построение дерева из `range`

```python
tree_range = BinaryTree()                   # создаём пустое бинарное дерево поиска
for i in range(5, 0, -1):                          # добавляем числа от 5 до 1 в обратном порядке
    tree_range.insert(i)                           # для чисел (int) определены операции сравнения
print(tree_range)
```

# Задание 7.4. Перегрузка операции принадлежности `in`

Переопределим классы `BinaryTree`, `BinaryNode`, `EmptyNode` за счёт определения нового метода `__contains__(self, value)` для перегрузки операции принадлежности `in`.

Операция `in`, вызванная для экземпляра `tree` класса `BinaryTree` в виде `value in tree`, будет возвращать `True`, если значение `value` найдено в дереве, и `False`, если значение не найдено.

```python
class EmptyNode:                          # класс для представления пустой вершины бинарного дерева поиска
    def __repr__(self):                            # метод строкового представления экземпляра класса
        return '*'                                 # пустая вершина выводится символом *
    def insert(self, value):                       # метод вставки в пустую вершину
        return BinaryNode(self, value, self)       # возвращаем новый экземпляр BinaryNode с пустыми поддеревьями
    def __contains__(self, value):                 # метод перегрузки операции in
        return False                               # в пустой вершине значение не найдено

class BinaryNode:                         # класс для представления вершины бинарного дерева поиска со значением и поддеревьями
    def __init__(self, left, value, right):        # метод инициализации экземпляра класса
        self.left = left                           # атрибут left — ссылка на корень левого поддерева
        self.value = value                         # атрибут value — значение вершины
        self.right = right                         # атрибут right — ссылка на корень правого поддерева
    def __repr__(self):                            # метод строкового представления экземпляра класса
        return f'({self.left}, {self.value}, {self.right})'  # возвращаем кортеж (левое поддерево, значение, правое поддерево)
    def insert(self, value):                       # метод вставки новой вершины
        if value < self.value:                     # если добавляемое значение меньше значения текущей вершины
            self.left = self.left.insert(value)    # рекурсивно вставляем в левое поддерево
        else:                                      # если добавляемое значение больше или равно
            self.right = self.right.insert(value)  # рекурсивно вставляем в правое поддерево
        return self                                # возвращаем ссылку на текущую вершину
    def __contains__(self, value):                 # метод перегрузки операции in
        if value == self.value:                    # если значение равно значению текущей вершины
            return True                            # значение найдено
        elif value < self.value:                   # если значение меньше значения текущей вершины
            return value in self.left              # рекурсивно ищем в левом поддереве
        else:                                      # если значение больше значения текущей вершины
            return value in self.right             # рекурсивно ищем в правом поддереве

class BinaryTree:                         # класс для представления бинарного дерева поиска
    def __init__(self):                            # метод инициализации экземпляра класса
        self.root = EmptyNode()                    # атрибут root — ссылка на корень дерева (изначально пустая вершина)
    def __repr__(self):                            # метод строкового представления экземпляра класса
        return repr(self.root)                     # возвращаем строковое представление корня
    def insert(self, value):                       # метод вставки элемента в дерево
        self.root = self.root.insert(value)        # переопределяем корень результатом вставки
    def __contains__(self, value):                 # метод перегрузки операции in
        return value in self.root                  # вызываем операцию in для корневой вершины
```

### Тестирование операции `in`

```python
source_data = [5, 1, 10, 3, 4]              # список числовых значений для построения дерева
tree = BinaryTree()                                # создаём пустое бинарное дерево поиска
for i in source_data:                              # последовательно вставляем элементы в дерево
    tree.insert(i)
    print(tree)                                    # выводим дерево после каждой вставки

for i in range(10):                                # проверяем принадлежность чисел от 0 до 9
    print((i, i in tree), end=' ')                 # (i, True) если i в дереве, (i, False) если нет
```

```python
tree_str = BinaryTree()                     # создаём пустое дерево
for ch in 'binary':                                # вставляем символы строки
    tree_str.insert(ch)
print('a' in tree_str, 'z' in tree_str)            # 'a' есть в дереве, 'z' нет
```

# Задание 7.5. Перегрузка встроенной функции `len`

Переопределим классы `BinaryTree`, `BinaryNode`, `EmptyNode` за счёт определения нового метода `__len__(self)` для перегрузки встроенной функции `len`, которая возвращает количество вершин в бинарном дереве поиска.

По лекции: метод `__len__` класса `BinaryNode` возвращает значение атрибута `numberOfNodes` класса `BinaryNode`. Атрибут `numberOfNodes` является атрибутом класса, а не атрибутом экземпляра. При создании класса атрибут `numberOfNodes` инициализируется значением 0. При создании экземпляра класса значение `numberOfNodes` увеличивается на 1. При создании дерева атрибут `numberOfNodes` «сбрасывается» в значение 0.

```python
class EmptyNode:                          # класс для представления пустой вершины бинарного дерева поиска
    def __repr__(self):                            # метод строкового представления экземпляра класса
        return '*'                                 # пустая вершина выводится символом *
    def insert(self, value):                       # метод вставки в пустую вершину
        return BinaryNode(self, value, self)       # возвращаем новый экземпляр BinaryNode с пустыми поддеревьями
    def __contains__(self, value):                 # метод перегрузки операции in
        return False                               # в пустой вершине значение не найдено
    def __len__(self):                             # метод перегрузки функции len
        return 0                                   # пустая вершина содержит 0 элементов

class BinaryNode:                         # класс для представления вершины бинарного дерева поиска со значением и поддеревьями
    numberOfNodes = 0                              # атрибут класса — счётчик созданных экземпляров (не атрибут экземпляра)
    def __init__(self, left, value, right):        # метод инициализации экземпляра класса
        self.left = left                           # атрибут left — ссылка на корень левого поддерева
        self.value = value                         # атрибут value — значение вершины
        self.right = right                         # атрибут right — ссылка на корень правого поддерева
        BinaryNode.numberOfNodes += 1              # при создании экземпляра увеличиваем счётчик на 1
    def __repr__(self):                            # метод строкового представления экземпляра класса
        return f'({self.left}, {self.value}, {self.right})'  # возвращаем кортеж (левое поддерево, значение, правое поддерево)
    def insert(self, value):                       # метод вставки новой вершины
        if value < self.value:                     # если добавляемое значение меньше значения текущей вершины
            self.left = self.left.insert(value)    # рекурсивно вставляем в левое поддерево
        else:                                      # если добавляемое значение больше или равно
            self.right = self.right.insert(value)  # рекурсивно вставляем в правое поддерево
        return self                                # возвращаем ссылку на текущую вершину
    def __contains__(self, value):                 # метод перегрузки операции in
        if value == self.value:                    # если значение равно значению текущей вершины
            return True                            # значение найдено
        elif value < self.value:                   # если значение меньше значения текущей вершины
            return value in self.left              # рекурсивно ищем в левом поддереве
        else:                                      # если значение больше значения текущей вершины
            return value in self.right             # рекурсивно ищем в правом поддереве
    def __len__(self):                             # метод перегрузки функции len
        return self.numberOfNodes                  # возвращаем значение атрибута класса numberOfNodes

class BinaryTree:                         # класс для представления бинарного дерева поиска
    def __init__(self):                            # метод инициализации экземпляра класса
        BinaryNode.numberOfNodes = 0               # сбрасываем счётчик при создании нового дерева
        self.root = EmptyNode()                    # атрибут root — ссылка на корень дерева (изначально пустая вершина)
    def __repr__(self):                            # метод строкового представления экземпляра класса
        return repr(self.root)                     # возвращаем строковое представление корня
    def insert(self, value):                       # метод вставки элемента в дерево
        self.root = self.root.insert(value)        # переопределяем корень результатом вставки
    def __contains__(self, value):                 # метод перегрузки операции in
        return value in self.root                  # вызываем операцию in для корневой вершины
    def __len__(self):                             # метод перегрузки функции len
        return len(self.root)                      # вызываем len для корневой вершины
```

### Тестирование функции `len`

```python
source_data = [5, 1, 10, 3, 4]              # список числовых значений для построения дерева
tree = BinaryTree()                                # создаём пустое бинарное дерево поиска
for i in source_data:                              # последовательно вставляем элементы в дерево
    tree.insert(i)
len(tree)                                          # количество вершин в дереве
```

### Пример 2: `len` для дерева из строки

```python
tree_str = BinaryTree()                     # создаём пустое дерево
for ch in 'binary':                                # вставляем символы строки
    tree_str.insert(ch)
len(tree_str)                                      # количество вершин в дереве
```

### Пример 3: `len` для дерева из range

```python
tree_range = BinaryTree()                   # создаём пустое дерево
for i in range(5, 0, -1):                          # вставляем числа от 5 до 1
    tree_range.insert(i)
len(tree_range)                                    # количество вершин в дереве
```

# Задание 7.6. Трассировка объектного уровня

Выполним трассировку процесса создания дерева по числовым данным `source_data = [5, 1, 10, 3, 4]`, следуя описанию вставки вершины в «пустое дерево» на Шаге 2 из лекционных материалов.

**Шаг 1 — Создание пустого дерева.** Экземпляр класса `BinaryTree` создаётся с атрибутом `root`, ссылающимся на экземпляр `EmptyNode`.

```python
tree = BinaryTree()                        # создаём пустое дерево; root = EmptyNode()
print(tree)                                        # пустое дерево выводится символом *
```

**Шаг 2 — Вставка значения 5.** Вызывается `tree.insert(5)`. Метод `insert` класса `BinaryTree` вызывает `self.root.insert(5)`. Поскольку `self.root` — экземпляр `EmptyNode`, вызывается метод `insert` класса `EmptyNode`, который возвращает новый экземпляр `BinaryNode(EmptyNode(), 5, EmptyNode())`. Атрибут `self.root` переопределяется ссылкой на новый объект.

```python
tree.insert(5)                             # вставляем 5: EmptyNode.insert(5) → BinaryNode(EmptyNode, 5, EmptyNode)
print(tree)                                        # дерево: (*, 5, *)
```

**Шаг 3 — Вставка значения 1.** Вызывается `tree.insert(1)`. Метод `insert` класса `BinaryTree` вызывает `self.root.insert(1)`. Поскольку `self.root` — экземпляр `BinaryNode` со значением 5, вызывается метод `insert` класса `BinaryNode`. Так как 1 < 5, вызывается `self.left.insert(1)`. `self.left` — экземпляр `EmptyNode`, который возвращает `BinaryNode(EmptyNode(), 1, EmptyNode())`. Атрибут `self.left` переопределяется.

```python
tree.insert(1)                             # 1 < 5 → идём влево: EmptyNode.insert(1) → BinaryNode(EmptyNode, 1, EmptyNode)
print(tree)                                        # дерево: ((*, 1, *), 5, *)
```

**Шаг 4 — Вставка значения 10.** Вызывается `tree.insert(10)`. Корень — `BinaryNode` со значением 5. Так как 10 >= 5, вызывается `self.right.insert(10)`. `self.right` — экземпляр `EmptyNode`, который возвращает `BinaryNode(EmptyNode(), 10, EmptyNode())`.

```python
tree.insert(10)                            # 10 >= 5 → идём вправо: EmptyNode.insert(10) → BinaryNode(EmptyNode, 10, EmptyNode)
print(tree)                                        # дерево: ((*, 1, *), 5, (*, 10, *))
```

**Шаг 5 — Вставка значения 3.** Вызывается `tree.insert(3)`. Корень — `BinaryNode` со значением 5. Так как 3 < 5, идём в левое поддерево (вершина со значением 1). Так как 3 >= 1, вызывается `self.right.insert(3)` для правого поддерева вершины 1. Правое поддерево — `EmptyNode`, возвращается `BinaryNode(EmptyNode(), 3, EmptyNode())`.

```python
tree.insert(3)                             # 3 < 5 → идём влево; 3 >= 1 → идём вправо: EmptyNode.insert(3)
print(tree)                                        # дерево: ((*, 1, (*, 3, *)), 5, (*, 10, *))
```

**Шаг 6 — Вставка значения 4.** Вызывается `tree.insert(4)`. Корень — `BinaryNode` со значением 5. Так как 4 < 5, идём в левое поддерево (вершина 1). Так как 4 >= 1, идём в правое поддерево (вершина 3). Так как 4 >= 3, вызывается `self.right.insert(4)`. Правое поддерево — `EmptyNode`, возвращается `BinaryNode(EmptyNode(), 4, EmptyNode())`.

```python
tree.insert(4)                             # 4 < 5 → влево; 4 >= 1 → вправо; 4 >= 3 → вправо: EmptyNode.insert(4)
print(tree)                                        # дерево: ((*, 1, (*, 3, (*, 4, *))), 5, (*, 10, *))
```
