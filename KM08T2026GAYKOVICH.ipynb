# Задание 8.1. Центрированный обход бинарного дерева поиска

Согласно лекции, центрированный обход (LCR) выполняется по правилу:

**левое поддерево → корень → правое поддерево**.

Результатом такого обхода для бинарного дерева поиска является список элементов, отсортированный по возрастанию.


## Класс EmptyNode

```python
class EmptyNode:
    # вставка элемента в пустую вершину
    def insert(self, value):
        # создаём новую вершину со значением value
        return BinaryNode(value)

    # центрированный обход пустой вершины
    def lcr(self):
        # для пустой вершины возвращается пустой список
        return []

    # поиск минимального значения
    def min(self):
        # если дерево пустое, возвращаем None
        return None

    # поиск максимального значения
    def max(self):
        # если дерево пустое, возвращаем None
        return None
```

Класс `EmptyNode` используется для представления пустого дерева и пустых поддеревьев.
Для пустой вершины центрированный обход возвращает пустой список.


## Класс BinaryNode

```python
class BinaryNode:
    # создание вершины дерева
    def __init__(self, value):
        # сохраняем значение вершины
        self.value = value

        # левое поддерево пока пустое
        self.left = EmptyNode()

        # правое поддерево пока пустое
        self.right = EmptyNode()

    # вставка нового элемента
    def insert(self, value):

        # если значение меньше текущего
        if value < self.value:

            # добавляем в левое поддерево
            self.left = self.left.insert(value)

        else:

            # иначе добавляем в правое поддерево
            self.right = self.right.insert(value)

        return self

    # центрированный обход дерева
    def lcr(self):

        # сначала обходим левое поддерево
        left_part = self.left.lcr()

        # затем текущую вершину
        center_part = [self.value]

        # затем правое поддерево
        right_part = self.right.lcr()

        # объединяем результаты
        return left_part + center_part + right_part

    # поиск минимума
    def min(self):

        # если слева пусто
        if isinstance(self.left, EmptyNode):

            # найден минимальный элемент
            return self.value

        # продолжаем двигаться влево
        return self.left.min()

    # поиск максимума
    def max(self):

        # если справа пусто
        if isinstance(self.right, EmptyNode):

            # найден максимальный элемент
            return self.value

        # продолжаем двигаться вправо
        return self.right.max()
```

Метод `lcr()` реализует центрированный обход дерева.

Сначала выполняется обход левого поддерева, затем добавляется значение текущей вершины, после чего выполняется обход правого поддерева.


## Класс BinaryTree

```python
class BinaryTree:

    # создание пустого дерева
    def __init__(self):

        # корень дерева пустой
        self.root = EmptyNode()

    # вставка нового элемента
    def insert(self, value):

        # вызываем вставку для корня
        self.root = self.root.insert(value)

    # центрированный обход
    def lcr(self):

        # запускаем обход от корня
        return self.root.lcr()

    # поиск минимума
    def min(self):

        # вызываем метод для корня
        return self.root.min()

    # поиск максимума
    def max(self):

        # вызываем метод для корня
        return self.root.max()
```

### Создание списка случайных чисел и построение дерева

```python
import random as rnd
import matplotlib.pyplot as plt

# создаём список из 1000 случайных чисел
source_data = [rnd.uniform(0, 100) for i in range(10**3)]

# создаём пустое дерево
tree = BinaryTree()

# добавляем все элементы списка в дерево
for i in source_data:
    tree.insert(i)
```

### Выполнение центрированного обхода

```python
# выполняем центрированный обход
sorted_data = tree.lcr()

# выводим первые элементы результата
print(sorted_data[:10])
```

После выполнения центрированного обхода получаем список элементов, отсортированный по возрастанию.


### Построение графика

```python
# график исходных данных
plt.plot(source_data,'r.',label='Initial data')

# график отсортированных данных
plt.plot(sorted_data,color='blue',label='Sorted data')

# заголовок графика
plt.title('Sequences of numbers')

# вывод легенды
plt.legend()

# отображение графика
plt.show()
```

# Задание 8.2. Нахождение минимального и максимального значения

Согласно лекции:

- для поиска минимального значения необходимо двигаться по левым поддеревьям;
- для поиска максимального значения необходимо двигаться по правым поддеревьям.


```python
# находим минимальное значение
print('Минимум =', tree.min())

# находим максимальное значение
print('Максимум =', tree.max())
```

Полученные значения являются минимальным и максимальным элементами дерева.

# Задание 8.4. Примеры пользовательских классов

В задании требуется привести три собственных класса с использованием возможностей ООП.


## Пример 1. Свойство property

```python
class Temperature:

    # создаём объект
    def __init__(self, value):

        # сохраняем температуру
        self._value = value

    @property
    def celsius(self):

        # возвращаем температуру
        return self._value


t = Temperature(25)

print(t.celsius)
```

## Пример 2. Классовый метод

```python
class Student:

    # счётчик объектов
    count = 0

    def __init__(self,name):

        self.name = name

        Student.count += 1

    @classmethod
    def total(cls):

        return cls.count


a = Student('Ivan')
b = Student('Petr')

print(Student.total())
```

## Пример 3. Абстрактный класс

```python
from abc import ABC, abstractmethod

class Shape(ABC):

    @abstractmethod
    def area(self):
        pass


class Square(Shape):

    def __init__(self,a):

        self.a = a

    def area(self):

        return self.a ** 2


sq = Square(5)

print(sq.area())
```
